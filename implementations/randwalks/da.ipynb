```python
import pandas as pd
import numpy as np
from numpy import genfromtxt
import matplotlib.pyplot as plt
```

```python
arr = genfromtxt('results.csv', delimiter=',')

data = arr[:-1]
# Create logarithmic bins
bins = np.logspace(np.log10(1), np.log10(2000), 400)  # 100 bins from 1 to 5000, logarithmically spaced
graph = np.histogram(data, bins)
likelihoods = [g / len(data) for g in graph[0]]
print(likelihoods)
print(graph[1][1:])
plt.figure(figsize=(10, 6))
plt.scatter(graph[1][1:], likelihoods, alpha=0.7, s=5)  # Using scatter instead of semilogx
plt.xscale('log')  # Set x-axis to logarithmic scale
plt.xlabel('Value (log scale)')
plt.ylabel('Likelihood')
plt.title('Histogram with Logarithmic Bins')
plt.grid(True, which="both", ls="-")

# Create a second figure for the line graph
plt.figure(figsize=(10, 6))
plt.plot(graph[1][1:], likelihoods, alpha=0.7, linewidth=1.5)
plt.xscale('log')  # Set x-axis to logarithmic scale
plt.xlabel('Value (log scale)')
plt.ylabel('Likelihood')
plt.title('Line Graph with Logarithmic Bins')
plt.grid(True, which="both", ls="-")
```
