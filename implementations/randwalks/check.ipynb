```python
import torch
import torch.nn as nn
import torch.nn.functional as F
from torchvision import datasets
```

```python
class NN(nn.Module):
    def __init__(self):
        super().__init__()
        self.l1 = nn.Linear(128, 10)
    
    def forward(self, x):
        x = F.relu(self.l1(x))
        return x
    

model = NN()

input = torch.randn(64, 128)
output = model(input)

print(output)
```
